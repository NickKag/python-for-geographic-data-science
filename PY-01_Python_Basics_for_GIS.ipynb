## 4. Αριθμητικοί υπολογισμοί

Η Python μπορεί να λειτουργήσει σαν αριθμομηχανή.

| Τελεστής | Σημασία |
|---|---|
| `+` | πρόσθεση |
| `-` | αφαίρεση |
| `*` | πολλαπλασιασμός |
| `/` | διαίρεση |
| `**` | δύναμη |
| `%` | υπόλοιπο διαίρεσης |


````python
population = 3150000
area_km2 = 2929

population_density = population / area_km2

print(population_density)
````

Η πυκνότητα πληθυσμού είναι:

$$
D = \frac{P}{A}
$$


όπου:

- \(D\) = πυκνότητα πληθυσμού,
- \(P\) = πληθυσμός,
- \(A\) = έκταση.

Αν η έκταση είναι σε km², το αποτέλεσμα είναι **κάτοικοι ανά km²**.


````python
population_density = round(population / area_km2, 2)

print(population_density)
````

## 5. Κείμενο και f-strings

Τα f-strings μάς επιτρέπουν να συνδυάζουμε κείμενο και μεταβλητές με καθαρό τρόπο.


````python
city = "Athens"
population_density = 1075.45

print(f"The population density of {city} is {population_density} people per km².")
````

Μπορούμε επίσης να μορφοποιήσουμε έναν αριθμό μέσα στο ίδιο f-string:


````python
city = "Athens"
population = 3150000
area_km2 = 2929
density = population / area_km2

print(f"{city}: {density:.2f} people/km²")
````

## 6. Συγκρίσεις και Boolean τιμές

Οι συγκρίσεις επιστρέφουν `True` ή `False`.

| Τελεστής | Σημασία |
|---|---|
| `==` | ίσο |
| `!=` | διαφορετικό |
| `>` | μεγαλύτερο |
| `<` | μικρότερο |
| `>=` | μεγαλύτερο ή ίσο |
| `<=` | μικρότερο ή ίσο |


````python
population = 3150000

print(population > 1000000)
print(population < 100000)
print(population == 3150000)
````

## 7. Μια πρώτη συνθήκη `if`

Η `if` μάς επιτρέπει να εκτελέσουμε κώδικα μόνο όταν μια συνθήκη είναι αληθής.


````python
population = 3150000

if population > 1000000:
    print("Large population")
````

Με `if` και `else` μπορούμε να επιλέξουμε ανάμεσα σε δύο περιπτώσεις.


````python
population = 85000

if population >= 100000:
    print("Large municipality")
else:
    print("Smaller municipality")
````

## 8. Απλές συναρτήσεις

Μια συνάρτηση είναι ένα μικρό επαναχρησιμοποιήσιμο κομμάτι κώδικα.

Αν θέλουμε να υπολογίζουμε πυκνότητα για πολλές περιοχές, δεν χρειάζεται να ξαναγράφουμε τον ίδιο τύπο κάθε φορά.


````python
def population_density(population, area_km2):
    return population / area_km2

density = population_density(3150000, 2929)

print(density)
````

Μπορούμε να χρησιμοποιήσουμε την ίδια συνάρτηση για οποιαδήποτε περιοχή:


````python
athens_density = population_density(3150000, 2929)
example_density = population_density(120000, 350)

print(f"Athens: {athens_density:.2f} people/km²")
print(f"Example area: {example_density:.2f} people/km²")
````
