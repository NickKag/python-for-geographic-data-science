## 1. Λίστες (`list`)

Αν θέλουμε να αποθηκεύσουμε πολλές πόλεις, χρησιμοποιούμε μια λίστα.


````python
cities = ["Athens", "Thessaloniki", "Patras", "Heraklion"]

print(cities)
print(type(cities))
````

Μια λίστα:

- γράφεται με αγκύλες `[]`,
- περιέχει πολλές τιμές,
- διατηρεί τη σειρά των στοιχείων,
- μπορεί να αλλάξει μετά τη δημιουργία της.


## 2. Indexing

Η Python ξεκινά την αρίθμηση από το **0**.

```text
Athens         → 0
Thessaloniki   → 1
Patras         → 2
Heraklion      → 3
```


````python
cities = ["Athens", "Thessaloniki", "Patras", "Heraklion"]

print(cities[0])
print(cities[1])
print(cities[-1])
````

Το `-1` σημαίνει το τελευταίο στοιχείο της λίστας.

## 3. Slicing

Με slicing επιλέγουμε ένα τμήμα της λίστας.


````python
cities = ["Athens", "Thessaloniki", "Patras", "Heraklion", "Larissa"]

print(cities[0:3])
print(cities[2:])
````

Στο `cities[0:3]` το στοιχείο με index `3` **δεν συμπεριλαμβάνεται**.

## 4. Προσθήκη και αλλαγή στοιχείων

````python
cities = ["Athens", "Thessaloniki", "Patras"]

cities.append("Larissa")
print(cities)

cities[2] = "Heraklion"
print(cities)
````

````python
cities = ["Athens", "Thessaloniki", "Patras", "Heraklion"]

print(len(cities))
print("Patras" in cities)
print("Volos" in cities)
````

## 5. Tuples

Ένα `tuple` μοιάζει με λίστα, αλλά συνήθως το χρησιμοποιούμε όταν οι τιμές δεν θέλουμε να αλλάζουν.

Παράδειγμα: ένα ζεύγος συντεταγμένων.


````python
coordinates = (23.7275, 37.9838)

longitude = coordinates[0]
latitude = coordinates[1]

print(longitude)
print(latitude)
print(type(coordinates))
````

Για αρχή θυμήσου:

- `list` → μεταβαλλόμενη συλλογή
- `tuple` → σταθερή συλλογή


## 6. Dictionaries (`dict`)

Ένα dictionary αποθηκεύει πληροφορίες ως:

```text
κλειδί → τιμή
```

Αυτό ταιριάζει πολύ καλά με γεωγραφικά αντικείμενα που έχουν πολλές ιδιότητες.


````python
city = {
    "name": "Athens",
    "population": 3150000,
    "area_km2": 2929
}

print(city["name"])
print(city["population"])
````

````python
city["country"] = "Greece"
print(city)
````

Ένα dictionary θυμίζει μια γραμμή πίνακα χαρακτηριστικών:

| name | population | area_km2 |
|---|---:|---:|
| Athens | 3150000 | 2929 |


# 7. `for` loops

Με `for` επαναλαμβάνουμε μια ενέργεια για κάθε στοιχείο μιας συλλογής.


````python
cities = ["Athens", "Thessaloniki", "Patras", "Heraklion"]

for city in cities:
    print(city)
````

Διάβασε τον κώδικα σαν πρόταση:

> Για κάθε `city` μέσα στη λίστα `cities`, εμφάνισε την πόλη.

Η εσοχή είναι σημαντική στην Python.


## 8. Υπολογισμοί μέσα σε loop

````python
populations = [3150000, 1100000, 215000, 175000]

for population in populations:
    population_thousands = population / 1000
    print(population_thousands)
````

## 9. `range()`

````python
for number in range(5):
    print(number)
````

Η `range(5)` παράγει:

```text
0
1
2
3
4
```

Σταματά πριν από το `5`.


````python
for number in range(1, 6):
    print(number)
````

## 10. `if`, `elif`, `else` μέσα σε loop

````python
populations = [45000, 120000, 650000, 18000]

for population in populations:
    if population >= 500000:
        print("Large urban population")
    elif population >= 100000:
        print("Medium population")
    else:
        print("Small population")
````

Το βασικό μοτίβο είναι:

```text
συλλογή δεδομένων
      ↓
for
      ↓
if / elif / else
      ↓
αποτέλεσμα
```


## 11. Λίστα από dictionaries

Τώρα συνδυάζουμε λίστες και dictionaries.


````python
cities = [
    {"name": "Athens", "population": 3150000, "area_km2": 2929},
    {"name": "Thessaloniki", "population": 1100000, "area_km2": 1455},
    {"name": "Patras", "population": 215000, "area_km2": 334}
]

print(cities[0])
print(cities[0]["name"])
````

````python
for city in cities:
    print(city["name"], city["population"])
````

# 12. Συναρτήσεις μέσα σε loops

Θυμόμαστε τη συνάρτηση πυκνότητας πληθυσμού από το PY-01:


````python
def population_density(population, area_km2):
    return population / area_km2
````

````python
for city in cities:
    density = population_density(
        city["population"],
        city["area_km2"]
    )

    print(f'{city["name"]}: {density:.2f} people/km²')
````

Το μοτίβο:

```text
δεδομένα → loop → function → αποτέλεσμα
```

θα το ξαναδούμε πολλές φορές αργότερα.

# Mini Project — Μικρή γεωγραφική ταξινόμηση

Για κάθε υποθετική πόλη:

1. υπολόγισε την πυκνότητα πληθυσμού,
2. ταξινόμησέ την:
   - `High density` αν `>= 1000`
   - `Medium density` αν `>= 300`
   - `Low density` διαφορετικά
3. εμφάνισε ένα καθαρό μήνυμα.

Παράδειγμα:

```text
GeoCity A — 597.56 people/km² — Medium density
```


````python
places = [
    {"name": "GeoCity A", "population": 245000, "area_km2": 410},
    {"name": "GeoCity B", "population": 850000, "area_km2": 520},
    {"name": "GeoCity C", "population": 90000, "area_km2": 650},
    {"name": "GeoCity D", "population": 430000, "area_km2": 300}
]

# Γράψε τον κώδικά σου εδώ
````

## Λύση Mini Project

Προσπάθησε πρώτα μόνος/η σου.


````python
def population_density(population, area_km2):
    return population / area_km2


for place in places:
    density = population_density(
        place["population"],
        place["area_km2"]
    )

    if density >= 1000:
        category = "High density"
    elif density >= 300:
        category = "Medium density"
    else:
        category = "Low density"

    print(
        f'{place["name"]} — '
        f'{density:.2f} people/km² — '
        f'{category}'
    )
````
