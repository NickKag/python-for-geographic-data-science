## 2. Από Python list σε NumPy array

Μία Python list μπορεί να αποθηκεύσει αριθμούς:

```python
elevations = [120, 135, 160, 180]
```

Το NumPy μας επιτρέπει να μετατρέψουμε τη λίστα σε αριθμητικό array:

````python
elevations = np.array([120 ,135, 160, 180])

print(elevations)
print(type(elevations))
````

Το αντικείμενο που δημιουργείται είναι `numpy.ndarray`.

Σε αντίθεση με μία απλή list, το NumPy array είναι σχεδιασμένο για γρήγορους αριθμητικούς υπολογισμούς.

## 3. Βασικά χαρακτηριστικά ενός array

Τρία πολύ χρήσιμα attributes είναι:

- `shape` → το σχήμα του array,
- `ndim` → ο αριθμός των διαστάσεων,
- `dtype` → ο τύπος των δεδομένων.

````python
print("Shape:", elevations.shape)
print("Dimensions:", elevations.ndim)
print("Data type:", elevations.dtype)
````

Για ένα μονοδιάστατο array με 4 τιμές, το `shape` είναι `(4,)`.

Το κόμμα δηλώνει ότι πρόκειται για tuple μίας διάστασης.

## 4. Indexing

Όπως στις lists, η αρίθμηση ξεκινά από το `0`.

````python
print(elevations[0])
print(elevations[1])
print(elevations[-1])
````

## 5. Slicing

Μπορούμε να πάρουμε ένα τμήμα του array:

````python
print(elevations[1:3])
print(elevations[:2])
print(elevations[2:])
````

Η έκφραση:

```python
elevations[1:3]
```

επιστρέφει τις θέσεις `1` και `2`, αλλά όχι τη θέση `3`.

## 6. Vectorized arithmetic

Εδώ βρίσκεται μία από τις σημαντικότερες διαφορές ανάμεσα σε lists και NumPy arrays.

Αν έχουμε:

````python
distances_km = np.array([2.5, 4.0, 7.2, 10.0])
````

και θέλουμε να μετατρέψουμε τα χιλιόμετρα σε μέτρα, δεν χρειαζόμαστε `for` loop:

````python
distances_m = distances_km * 1000

print(distances_m)
````

Το NumPy εφαρμόζει την πράξη σε **κάθε στοιχείο του array**.

Αυτό ονομάζεται συνήθως **vectorized operation**.

### Παράδειγμα: μεταβολή υψομέτρου

````python
elevations = np.array([120, 135, 160, 180])

elevations_plus_10 = elevations + 10

print(elevations_plus_10)
````

## 7. Πράξεις μεταξύ arrays

Αν δύο arrays έχουν συμβατό σχήμα, μπορούμε να κάνουμε πράξεις στοιχείο προς στοιχείο.

````python
x = np.array([10, 20, 30])
y = np.array([1, 2, 3])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
````

## 8. Βασικά στατιστικά

Το NumPy μπορεί να υπολογίσει γρήγορα βασικά στατιστικά μέτρα.

````python
elevations = np.array([120, 135, 160, 180, 210, 95])

print("Minimum:", elevations.min())
print("Maximum:", elevations.max())
print("Mean:", elevations.mean())
print("Standard deviation:", elevations.std())
````

Μπορούμε επίσης να χρησιμοποιήσουμε τις αντίστοιχες συναρτήσεις του NumPy:

````python
print(np.min(elevations))
print(np.max(elevations))
print(np.mean(elevations))
print(np.std(elevations))
````

## 9. Boolean masks

Ένα Boolean mask είναι ένας πίνακας από `True` και `False`.

Για παράδειγμα:

````python
elevations = np.array([120, 135, 160, 180, 210, 95])

mask = elevations >= 150

print(mask)
````

Το αποτέλεσμα δείχνει ποιες τιμές ικανοποιούν τη συνθήκη.

Μπορούμε να χρησιμοποιήσουμε το mask για φιλτράρισμα:

````python
high_elevations = elevations[elevations >= 150]

print(high_elevations)
````

Αυτό είναι πολύ σημαντικό για GIS workflows.

Για παράδειγμα, αργότερα μπορούμε να κάνουμε κάτι όπως:

```python
suitable = slope < 15
```

ή:

```python
high_ndvi = ndvi > 0.5
```

## 10. Δισδιάστατα arrays

Τα raster δεδομένα μπορούν να τα φανταστούμε ως πλέγματα από αριθμούς.

Για παράδειγμα:

````python
dem = np.array([
    [100, 105, 110],
    [120, 125, 130],
    [140, 145, 150]
])

print(dem)
````

Αυτό είναι ένα array δύο διαστάσεων.

````python
print("Shape:", dem.shape)
print("Dimensions:", dem.ndim)
````

Το `shape` είναι:

```text
(3, 3)
```

δηλαδή 3 γραμμές και 3 στήλες.

## 11. Indexing σε δύο διαστάσεις

Η γενική μορφή είναι:

```python
array[row, column]
```

````python
print(dem[0, 0])
print(dem[1, 2])
print(dem[2, 1])
````

## 12. Slicing σε δύο διαστάσεις

Μπορούμε να επιλέξουμε ολόκληρες γραμμές ή στήλες.

````python
print("First row:")
print(dem[0, :])

print("Second column:")
print(dem[:, 1])
````

Το `:` σημαίνει ουσιαστικά «όλες οι θέσεις» σε εκείνη τη διάσταση.

## 13. Raster-like operations

Ας δημιουργήσουμε ένα απλό array που αντιπροσωπεύει κλίσεις σε μοίρες:

````python
slope = np.array([
    [5, 12, 25],
    [8, 18, 32],
    [3, 10, 20]
])

print(slope)
````

Θέλουμε να βρούμε ποια κελιά έχουν κλίση μικρότερη από 15°:

````python
suitable_slope = slope < 15

print(suitable_slope)
````

Μπορούμε επίσης να μετατρέψουμε το Boolean αποτέλεσμα σε `0` και `1`:

````python
suitable_slope_binary = (slope < 15).astype(int)

print(suitable_slope_binary)
````

Αυτό θυμίζει άμεσα ένα **constraint raster**:

- `1` → επιτρέπεται,
- `0` → αποκλείεται.

## 14. Συνδυασμός συνθηκών

Μπορούμε να συνδυάσουμε δύο NumPy συνθήκες με:

- `&` → AND
- `|` → OR

Χρησιμοποιούμε παρενθέσεις γύρω από κάθε συνθήκη.

````python
elevation = np.array([100, 250, 400, 550, 700])
slope = np.array([5, 12, 8, 20, 7])

suitable = (elevation < 600) & (slope < 15)

print(suitable)
````

Και μπορούμε να δούμε τις τιμές υψομέτρου που ικανοποιούν και τις δύο συνθήκες:

````python
print(elevation[suitable])
````

## 15. `NaN` — missing numerical values

Σε πραγματικά δεδομένα μπορεί να υπάρχουν τιμές που λείπουν.

Το NumPy χρησιμοποιεί συχνά το `np.nan`.

````python
measurements = np.array([12.5, 13.2, np.nan, 11.8, 14.0])

print(measurements)
````

Αν χρησιμοποιήσουμε το απλό `mean()`, το αποτέλεσμα θα είναι `nan`:

````python
print(measurements.mean())
````

Για να αγνοήσουμε προσωρινά τα `NaN`, υπάρχουν συναρτήσεις όπως:

````python
print(np.nanmean(measurements))
print(np.nanmin(measurements))
print(np.nanmax(measurements))
````

> Προσοχή: το ότι μπορούμε να αγνοήσουμε ένα missing value δεν σημαίνει ότι πρέπει πάντα να το κάνουμε. Αργότερα, στην επεξεργασία πραγματικών δεδομένων, θα εξετάζουμε πρώτα **γιατί** λείπει μία τιμή.

## 16. `reshape()`

Μερικές φορές θέλουμε να αλλάξουμε το σχήμα ενός array χωρίς να αλλάξουμε τις τιμές του.

````python
values = np.array([1, 2, 3, 4, 5, 6])

grid = values.reshape(2, 3)

print(grid)
````

Το αρχικό array είχε 6 στοιχεία.

Το νέο array έχει:

- 2 γραμμές,
- 3 στήλες,

και πάλι συνολικά 6 στοιχεία.

## 17. Mini GIS example — υψόμετρα σημείων

Έστω ότι έχουμε έξι σημεία με υψόμετρο σε μέτρα:

````python
elevations = np.array([85, 220, 410, 620, 150, 330])

print(elevations)
````

Θέλουμε να:

1. βρούμε το μέσο υψόμετρο,
2. εντοπίσουμε τα σημεία πάνω από 300 m,
3. υπολογίσουμε πόσα είναι,
4. υπολογίσουμε τι ποσοστό του συνόλου αντιπροσωπεύουν.

````python
mean_elevation = elevations.mean()

high_mask = elevations > 300
high_elevations = elevations[high_mask]

count_high = high_mask.sum()
percentage_high = count_high / elevations.size * 100

print(f"Mean elevation: {mean_elevation:.1f} m")
print(f"Elevations above 300 m: {high_elevations}")
print(f"Number of points above 300 m: {count_high}")
print(f"Percentage above 300 m: {percentage_high:.1f}%")
````

### Τι κάνει το `high_mask.sum()`;

Σε αριθμητικό context:

- `True` συμπεριφέρεται σαν `1`,
- `False` συμπεριφέρεται σαν `0`.

Άρα το άθροισμα ενός Boolean array μετρά πόσες φορές εμφανίζεται το `True`.
