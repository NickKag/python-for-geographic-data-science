## 2. Pandas Series

Ένα `Series` είναι μία μονοδιάστατη δομή δεδομένων.

Μπορούμε να το φανταστούμε σαν μία στήλη ενός πίνακα.

````python
populations = pd.Series([45000, 120000, 650000, 18000])

print(populations)
````

Παρατήρησε ότι το Pandas δημιουργεί αυτόματα έναν **index**:

```text

```

Κάθε τιμή έχει επομένως:

## 3. DataFrame

Το `DataFrame` είναι η βασική δομή του Pandas.

Μοιάζει με πίνακα Excel:

- κάθε **γραμμή** είναι μία παρατήρηση,
- κάθε **στήλη** είναι μία μεταβλητή.

````python
data = {
    "municipality": ["Alexandroupoli", "Komotini", "Xanthi", "Drama", "Kavala"],
    "population": [71959, 66919, 66487, 55482, 66576],
    "area_km2": [1217.0, 644.9, 495.1, 840.1, 350.6],
    "region": [
        "Evros",
        "Rodopi",
        "Xanthi",
        "Drama",
        "Kavala"
    ]
}

df = pd.DataFrame(data)

df
````

## 9. Filtering με Boolean συνθήκες

Αυτό δημιουργεί ένα Boolean Series.

## 10. Πολλαπλές συνθήκες

Στο Pandas χρησιμοποιούμε:

- `&` για AND
- `|` για OR

και βάζουμε κάθε συνθήκη σε παρενθέσεις.

````python
filtered = df[
    (df["population"] > 60000) &
    (df["area_km2"] < 700)
]

filtered
````

Με OR:

````python
filtered = df[
    (df["population"] > 70000) |
    (df["area_km2"] < 400)
]

filtered
````

## 11. Δημιουργία νέας στήλης

Στη Γεωγραφική Επιστήμη Δεδομένων συχνά δεν χρησιμοποιούμε μόνο τις αρχικές μεταβλητές.

Δημιουργούμε **δείκτες** από αυτές.

Για παράδειγμα, πληθυσμιακή πυκνότητα:

density = population / area

````python
df["population_density"] = (
    df["population"] / df["area_km2"]
)

df
````

Η νέα στήλη δημιουργείται χωρίς `for` loop.

Το Pandas εφαρμόζει την πράξη στις αντίστοιχες τιμές κάθε γραμμής.

````python
df[["municipality", "population_density"]]
````

Μπορούμε να στρογγυλοποιήσουμε τις τιμές:

````python
df["population_density"] = df["population_density"].round(2)

df[["municipality", "population_density"]]
````

## 15. `groupby()`

Το `groupby()` είναι μία από τις σημαντικότερες λειτουργίες του Pandas.

Ας χρησιμοποιήσουμε ένα μικρό dataset με περισσότερους οικισμούς ανά περιφερειακή ενότητα:

````python
settlements = pd.DataFrame({
    "settlement": [
        "A", "B", "C", "D", "E", "F"
    ],
    "regional_unit": [
        "Evros", "Evros", "Rodopi",
        "Rodopi", "Xanthi", "Xanthi"
    ],
    "population": [
        12000, 8000, 15000, 5000, 11000, 9000
    ]
})

settlements
````

Συνολικός πληθυσμός ανά περιφερειακή ενότητα:

````python
settlements.groupby("regional_unit")["population"].sum()
````

Μέσος πληθυσμός:

````python
settlements.groupby("regional_unit")["population"].mean()
````

Η λογική είναι:

```text
χωρίζω τις γραμμές σε ομάδες
        ↓
εφαρμόζω μία συνάρτηση
        ↓
παίρνω αποτέλεσμα ανά ομάδα
```

Αυτό συχνά περιγράφεται ως **split → apply → combine**.

## 16. Missing values

## 20. Μικρό GIS workflow

Ας συνδυάσουμε όσα μάθαμε.

Θέλουμε να:

1. υπολογίσουμε πληθυσμιακή πυκνότητα,
2. κρατήσουμε τους Δήμους με πυκνότητα πάνω από 100 κατοίκους/km²,
3. ταξινομήσουμε από μεγαλύτερη προς μικρότερη πυκνότητα,
4. εμφανίσουμε μόνο το όνομα και την πυκνότητα.

````python
municipalities = df.copy()

municipalities["population_density"] = (
    municipalities["population"] /
    municipalities["area_km2"]
)

result = municipalities[
    municipalities["population_density"] > 100
].sort_values(
    by="population_density",
    ascending=False
)

result[["municipality", "population_density"]]
````

Αυτή η ακολουθία είναι ήδη ένα μικρό **data-processing pipeline**:

```text
raw table
   ↓
create indicator
   ↓
filter
   ↓
sort
   ↓
select output columns
```

Στα επόμενα μαθήματα αυτή η λογική θα εφαρμοστεί σε πραγματικά στατιστικά και γεωγραφικά δεδομένα.
